```python
# ## What is Exploratory Data Analysis (EDA)?
# 
# Think of EDA as getting to know your data inside out before you build any machine learning model. It is a step-by-step process of checking, cleaning, and visualizing your data so you understand what you are working with.
# 
# ### Key Steps in EDA
# * **Load the Data:** Import your dataset into your environment (using tools like **Pandas**).
# * **Clean the Data:** Spot and fix missing values, errors, or messy formatting.
# * **Visualize the Data:** Draw charts and graphs (using libraries like **Matplotlib** and **Seaborn**) to discover patterns, trends, and outliers.
# 
# ### Why Does It Matter?
# Skipping EDA is like driving blind. Doing this foundational groundwork ensures your data is reliable before you feed it into a machine learning model.
```

```python
#Handling missing data:
#In real-world datasets,missing values are common, and is crucial to address them appropriately before conducting further analysis.
import pandas as pd
data = pd.read_csv('MBA.csv')
print(data.head())
print(data.isnull().sum())
```

```python
#How to handle missing data:
#There are two primary options for addressing missing values: dropping them or filling them with specific values.
#The first option is to drop missing values (using the dropna() method), which removes rows and columns containing missing data.
import pandas as pd
data = pd.read_csv('MBA.csv')
#data_clean = data.dropna()
#print(data_clean)
#print(data_clean.isnull().sum())
#The second option is to fill missing values by replacing them (using the fillna() method) with specific values, such as the mean, median, or a placeholder.
data_clean2 = data.fillna({'work_exp':data['work_exp'].mean()}, inplace=True)
print(data_clean2.isnull().sum())

#Note: For dropna() & fillna() methods comment & uncomment the respective method you want to run.
#Use dropna() when missing values represent less than 5 percent of your dataset
#Use fillna() when missing values represent more than 5 percent of your dataset
```

```python
#Feature Scaling 
#Feature scaling brings different features onto a comparable scale (such as 0 to 1 for normalization, or a mean of 0 and variance of 1 for standardization). It restricts or normalizes the range rather than expanding it multifold.
#This ensures that features with larger numerical ranges (like salary in lakhs) don't unfairly dominate features with smaller ranges (like experience in years) during optimization algorithms like gradient descent or distance-based models like KNN and SVM.
import pandas as pd 
data = pd.read_csv('data.csv')
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
data[['Age','Income']] = scaler.fit_transform(data[['Age','Income']])
print(data)
data_encoded =  pd.get_dummies(data,columns=['Age','Income'])
print(data_encoded)
```

```python
#Matplotlib
import pandas as pd
data = pd.read_csv('data.csv')
import matplotlib.pyplot as plt 
plt.plot(data['Age'],data['Income']) # A line plot for age vs income.
plt.title('Age vs Income') # A title to the plot.
plt.xlabel('Age') # Labelling x-axis.
plt.ylabel('Income') # Labelling y-axis.
plt.show()

#Bar chart
data['Gender'].value_counts().plot(kind='bar')
plt.title('Gender Distribution')
plt.show()
```

```python
# Hands-on project:
# To perform exploratory data analysis on the Titanic dataset to understand the survival rate across various features, including age, gender, and class

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 

data = pd.read_csv('Titanic Dataset.csv')
print(data.head())

#handling missing values in the dataset & focus is on Age.

data.fillna({'Age':data['Age'].mean()}, inplace=True)
print(data.isnull().sum())

# Analyse the survival rate: The following code prints the normalized value counts for the 'Survived' column.
#In Pandas, setting normalize=True inside the value_counts() method returns the relative frequencies (proportions) of unique values instead of their raw counts. The resulting values range between 0 and 1, and their total sum equals 1.0.

print(data['Survived'].value_counts(normalize=True))

#Group the data by gender and passenger class to analyze survival rates across these categories. The code groups the data by 'Sex' and 'Pclass', then calculates the mean of the 'survived' column for each group:

print(data.groupby(['Age','Pclass'])['Survived'].mean())

#Visualise the data:

sns.barplot(x='Sex', y='Survived', data=data)
plt.title('Survival Rate by Gender')
plt.show()
```
